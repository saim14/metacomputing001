Read-only review and reconstruction by default. Full reruns go to a new directory. No GPU required.


```python
from pathlib import Path
import subprocess
repo = Path("/content/metacomputing001")
if not repo.exists():
    subprocess.run(["git", "clone", "--depth", "1", "--filter=blob:none", "--sparse", "https://github.com/saim14/metacomputing001.git", str(repo)], check=True)
subprocess.run(["git", "-C", str(repo), "sparse-checkout", "set", "research002/experiments/F1_A3"], check=True)
study_dir = repo / "research002/experiments/F1_A3/main_v1"
```

## Optional: fresh deterministic reproduction
Changing seeds or hypotheses requires a new protocol. This reproduces the recorded study, not independent confirmation. Set RUN_FRESH to True; the destination must not exist.


```python
RUN_FRESH = False
if RUN_FRESH:
    import shutil
    fresh = Path("/content/f1_a3_reproduction")
    fresh.mkdir(exist_ok=False)
    for name in ["study.py", "config.json", "F1_A3_Protocol.md", "audit.py"]:
        shutil.copyfile(study_dir / name, fresh / name)
    shutil.copytree(study_dir / "deps", fresh / "deps")
    subprocess.run(["python", "study.py"], cwd=fresh, check=True)
    subprocess.run(["python", "audit.py"], cwd=fresh, check=True)
```
